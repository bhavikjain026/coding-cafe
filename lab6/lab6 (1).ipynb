## 1. Lists

**What it is:** A **list** is an ordered collection of items. Items stay in the order you add them, duplicates are allowed, and you can change the list after creating it.

Each item has a position number called an **index**. The first item is at index `0`, the second at index `1`, and so on. Index `-1` refers to the last item.

For example, `orders = ["chai", "samosa", "chai"]` has three items, and `orders[0]` is `"chai"`.

| Method or operation | What it does |
| --- | --- |
| `len(orders)` | Count the items |
| `orders[i]` | Read the item at position `i` (`orders[-1]` is the last item) |
| `orders[start:stop]` | **Slice:** copy the items from position `start` up to, but not including, position `stop` |
| `orders.append(x)` | Add `x` to the end |
| `orders.insert(i, x)` | Add `x` at position `i`. Items from position `i` onward move one place to the right |
| `orders.extend(items)` | Add every item from `items` to the end |
| `orders.remove(x)` | Delete the first `x`. Causes an error if `x` is not in the list |
| `orders.pop(i)` | Delete the item at position `i` and return it. With no number, it deletes the last item |
| `orders.count(x)` | Count how many times `x` appears |
| `orders.index(x)` | Return the position of the first `x` |
| `orders.sort()` | Sort the items: A to Z for text, smallest to largest for numbers |
| `orders.reverse()` | Reverse the order of the items |
| `orders.copy()` | Make a separate copy of the list |
| `orders.clear()` | Delete all items |

**In place:** `sort()`, `reverse()`, `append()`, and the other methods above change the list itself. They return `None`, so write `orders.sort()` on its own line. Do not write `new = orders.sort()`.

## 2. Sets

**What it is:** A **set** is an unordered collection of unique items. If you add duplicates, only one copy is kept. Sets have no index numbers, so `guests[0]` does not work. Use `in` to check whether an item is in a set.

Sets suit guest lists. The festival has two areas, **Lakeside** and **Hilltop**. Guests often book twice, and a set removes the repeated names automatically.

Four set operations come from Venn diagrams, which you may have seen in high school math:

- **Union (A ∪ B):** everyone in A, in B, or in both.
- **Intersection (A ∩ B):** everyone in both A and B.
- **Difference (A − B):** everyone in A who is not in B.
- **Symmetric difference (A △ B):** everyone in exactly one of A or B.

| Method or operation | What it does | Symbol version |
| --- | --- | --- |
| `set(items)` | Make a set from a list or another collection | — |
| `a.add(x)` | Add one item | — |
| `a.update(b)` | Add every item from `b` | `a \|= b` |
| `a.remove(x)` | Delete `x`. Causes an error if `x` is missing | — |
| `a.discard(x)` | Delete `x`. No error if `x` is missing | — |
| `a.clear()` | Delete all items | — |
| `a.copy()` | Make a separate copy | — |
| `x in a` | Check whether `x` is in `a` | — |
| `len(a)` | Count the items | — |
| `a.union(b)` | Items in A, in B, or in both (A ∪ B) | `a \| b` |
| `a.intersection(b)` | Items in both A and B (A ∩ B) | `a & b` |
| `a.difference(b)` | Items in A but not in B (A − B) | `a - b` |
| `a.symmetric_difference(b)` | Items in exactly one of A or B (A △ B) | `a ^ b` |
| `a.issubset(b)` | True if every item in A is also in B (A ⊆ B) | `a <= b` |
| `a.isdisjoint(b)` | True if A and B share no items | — |

The symbol versions give the same result as the named methods. This lab uses the named methods so each line explains itself.

**Note:** `{}` makes an empty dictionary, not a set. Write `set()` for an empty set.


### Worked example

Read the code and predict each printed line. Then run the cell to compare.


````python
# Remove duplicates, then compare two guest lists
morning = set(["Asha", "Ravi", "Asha", "Meera"])    # "Asha" is kept once
evening = {"Ravi", "Kabir", "Meera", "Tara"}

print("Morning guests:", sorted(morning))
print("Union:", sorted(morning.union(evening)))
print("Intersection:", sorted(morning.intersection(evening)))
print("Difference:", sorted(morning.difference(evening)))
print("Symmetric difference:", sorted(morning ^ evening))    # symbol version of symmetric_difference
print("Is 'Zoya' in morning?", "Zoya" in morning)
````

## 3. Tuples

**What it is:** A **tuple** is an ordered collection, like a list, but it cannot be changed after you create it. This is called being **immutable**. Tuples suit records with fixed fields. A recipe, for example, always has a name, a price, and a preparation time.

You read tuple items by index, the same way as lists: `recipe[0]` is the first item.

Tuples have only two methods. The table also lists the common tuple operations.

| Method or operation | What it does |
| --- | --- |
| `recipe = ("Masala Chai", 40, 4)` | Create a tuple with three fields |
| `recipe[0]`, `recipe[-1]` | Read the first or last item |
| `name, price, prep = recipe` | **Unpacking:** put each field into its own variable |
| `len(recipe)` | Count the items |
| `x in recipe` | Check whether `x` is one of the items |
| `recipe.count(x)` | Count how many times `x` appears |
| `recipe.index(x)` | Return the position of the first `x` |

## 4. Dictionaries

**What it is:** A **dictionary** (`dict`) stores **key-value pairs**. A key is a label, and its value is the data stored under that label. For example, the key `"Chai"` can hold the value `40`. You look up a value by its key, not by its position.

Keys must be unique. Values can be any type. Dictionaries can be changed, so you can add, change, and delete pairs. `{}` makes an empty dictionary.

| Method or operation | What it does |
| --- | --- |
| `prices[key]` | Read a value. Causes a `KeyError` if the key is missing |
| `prices.get(key, default)` | Read a value. Returns `default` if the key is missing |
| `prices[key] = value` | Add a new pair, or change the value of an existing key |
| `prices.update(other)` | Add or change several pairs at once, using another dictionary |
| `prices.setdefault(key, default)` | Add `key` with `default` only if it is missing. Returns its value |
| `prices.pop(key)` | Delete a pair and return its value |
| `prices.keys()` | Get all keys |
| `prices.values()` | Get all values |
| `prices.items()` | Get all `(key, value)` pairs, for use in a loop |
| `key in prices` | Check whether a key exists |
| `len(prices)` | Count the pairs |
| `prices.copy()` | Make a separate copy |
| `prices.clear()` | Delete all pairs |

**Note:** `keys()`, `values()`, and `items()` return special views. Wrap them in `list()` to print them as plain lists.


### Worked example

Read the code and predict each printed line. Then run the cell to compare.


````python
# Safe lookups, adding and changing pairs, and looping
menu_prices = {"Chai": 40, "Samosa": 30, "Bun Maska": 50}

print("Chai price:", menu_prices["Chai"])
print("Cold coffee price:", menu_prices.get("Cold Coffee", 0))    # missing key returns 0, with no error
menu_prices["Cold Coffee"] = 70      # add a new key
menu_prices["Chai"] = 45             # change an existing key

for item, price in menu_prices.items():
    print(f"{item}: Rs {price}")
````

### Your turn: Process a festival order

Use the prices, stock, and order below. Write the program in the cell below.

1. Add `"Lemonade"` to `prices` at Rs 60. Change the Samosa price to 35. Then delete `"Lemonade"` with `pop()`.
2. Loop over `order.items()`. Add each item's price times its quantity to a variable named `bill`.
3. If `bill` is more than 300, multiply it by `0.9` (a 10% discount) and store the result in `payable`. Otherwise, set `payable` equal to `bill`.
4. Create an empty list named `unavailable`. Loop over `order.items()`. If the quantity is more than `stock.get(item, 0)`, add the item to `unavailable`. Otherwise, subtract the quantity from `stock[item]`.
5. Print `bill`, `payable`, `unavailable`, and `stock`.

**Checkpoint:** Your output should show:
- Bill: `370`
- Payable: `333.0`
- Unavailable: `['Veg Puff']`

````python
prices = {"Masala Chai": 40, "Samosa": 30, "Chocolate Brownie": 90, "Veg Puff": 45}
stock = {"Masala Chai": 10, "Samosa": 4, "Chocolate Brownie": 8, "Veg Puff": 0}
order = {"Masala Chai": 3, "Samosa": 2, "Chocolate Brownie": 1, "Veg Puff": 2}

# 1. Update prices
prices["Lemonade"] = 60
prices["Samosa"] = 35
prices.pop("Lemonade")

# 2. Calculate the bill
bill = 0
for item, qty in order.items():
    bill += prices[item] * qty

# 3. Apply discount
if bill > 300:
    payable = bill * 0.9
else:
    payable = bill

# 4. Check stock availability and update stock
unavailable = []
for item, qty in order.items():
    if qty > stock.get(item, 0):
        unavailable.append(item)
    else:
        stock[item] -= qty

# 5. Print results
print("Bill:", bill)
print("Payable:", payable)
print("Unavailable:", unavailable)
print("Stock:", stock)
````

## 5. Reading CSV files line by line

**What it is:** A **CSV** file is a plain text file. Each line is one row, and commas separate the values in that row. The first line holds the column names.

For example, this line from `data/menu.csv` has five values:

`Masala Chai,Drinks,40,4,yes`

### How `split(",")` works

`split(",")` cuts a line of text into a **list** of pieces. It cuts at every comma, and the commas disappear.

- `"Masala Chai,Drinks,40".split(",")` gives `['Masala Chai', 'Drinks', '40']`.
- Every piece is text, even `'40'`. Use `int('40')` to turn it into the number 40.
- Two commas with nothing between them give an empty text, `''`. So `"a,,b".split(",")` gives `['a', '', 'b']`.
- Each line from a file ends with an invisible newline character. Use `.strip()` first to remove it.

| Code | What it does |
| --- | --- |
| `with open("data/menu.csv") as f:` | Open the file. `f` now stands for the open file |
| `next(f)` | Read one line, such as the header line |
| `for line in f:` | Visit one line at a time |
| `line.strip().split(",")` | Remove the line ending, then cut the line at the `','` into a list of values |
| `int(text)` | Turn text into a whole number |


### Worked example

Read the code and predict each printed line. Then run the cell to compare.


````python
# Read menu.csv into a dictionary: item name -> price
menu_prices = {}
with open("data/menu.csv") as f:
    next(f)                                  # skip the header line
    for line in f:
        fields = line.strip().split(",")     # e.g. ['Masala Chai', 'Drinks', '40', '4', 'yes']
        item, category, price, prep_min, vegetarian = fields # unpacks each item of the list into a variable
        menu_prices[item] = int(price)

print("Items loaded:", len(menu_prices))
print("Masala Chai price:", menu_prices["Masala Chai"])
print("Filter Coffee price:", menu_prices["Filter Coffee"])
````

## 7. Capstone: Loyalty list

To thank our regulars, Meera wants to give Coding Café Loyalty Cards to every customer who placed **two or more distinct orders** during the festival.

This task combines dictionaries, sets, tuples, and lists. Write your solution in the cell below. There are no blanks in this task.

1. Read both CSV files as in the worked example and the guided practice. For each sales row, work out `revenue = qty * price`.
2. Skip rows where `customer_id` is empty. For each customer, track three things: a set of `order_id` values, a set of `item` names, and the total spend. (Hint: `customers.setdefault(customer, {"orders": set(), "items": set(), "spend": 0})` creates an entry the first time a customer appears.)
3. Count the customers with at least 2 distinct orders. Print that count. Then print the customer ID with the highest spend, and that amount.
4. Use `intersection()` and `difference()` on the item sets of `C01` and `C02`. Print the items both customers bought, and the items only `C01` bought. Print each list sorted.
5. Build a list of tuples: `(customer_id, order_count, item_count, spend)`. Sort it by spend, highest first. Customers with the same spend are sorted by customer ID, A to Z. Print the top five.

*Sorting hint:* `cards.sort(key=lambda row: (-row[3], row[0]))` sorts by spend from highest to lowest, using the minus sign on `row[3]`. Ties are sorted by `row[0]`, the customer ID.

**Checkpoint:** There are 17 qualifying customers. The highest spender is `C04` with Rs 710. Items bought by both `C01` and `C02`: `['Banana Bread', 'Filter Coffee', 'Masala Chai', 'Samosa']`. Items only `C01` bought: `['Veg Puff']`. The top five are:

| customer_id | order_count | item_count | spend |
| --- | --- | --- | --- |

| C02 | 3 | 5 | 590 |
| C06 | 2 | 3 | 590 |
| C03 | 3 | 4 | 540 |
| C11 | 2 | 2 | 540 |


````python
# Load menu prices.
menu_prices = {}
with open("data/menu.csv") as f:
    next(f)
    for line in f:
        item, category, price, prep_min, vegetarian = line.strip().split(",")
        menu_prices[item] = int(price)

# Track distinct orders, distinct items, and total spend for each customer.
customers = {}
with open("data/monsoon_sales.csv") as f:
    columns = next(f).strip().split(",")
    for line in f:
        row = dict(zip(columns, line.strip().split(",")))
        customer_id = row["customer_id"].strip()
        if not customer_id:
            continue

        item = row["item"]
        qty = int(row["qty"])
        customer = customers.setdefault(
            customer_id, {"orders": set(), "items": set(), "spend": 0}
        )
        customer["orders"].add(row["order_id"])
        customer["items"].add(item)
        customer["spend"] += qty * menu_prices[item]

# Count customers who placed at least two distinct orders.
qualifying_ids = [
    customer_id
    for customer_id, details in customers.items()
    if len(details["orders"]) >= 2
]
print("Customers with 2+ orders:", len(qualifying_ids))

# Find the highest spender across customers with recorded sales.
highest_spender = max(customers, key=lambda cid: customers[cid]["spend"])
print("Highest spender:", highest_spender, "Rs", customers[highest_spender]["spend"])

# Compare the items bought by C01 and C02.
c01_items = customers["C01"]["items"]
c02_items = customers["C02"]["items"]
print("Items bought by both C01 and C02:", sorted(c01_items.intersection(c02_items)))
print("Items only C01 bought:", sorted(c01_items.difference(c02_items)))

# Build and rank loyalty-card tuples for qualifying customers.
cards = []
for customer_id in qualifying_ids:
    details = customers[customer_id]
    cards.append(
        (customer_id, len(details["orders"]), len(details["items"]), details["spend"])
    )
cards.sort(key=lambda row: (-row[3], row[0]))
print("Top five loyalty cards:")
for card in cards[:5]:
    print(card)
````
